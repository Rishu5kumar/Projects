# Simple Linear Regression

## Importing the libraries

```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

## Importing the dataset

```python
dataset=pd.read_csv("Salary.csv")
x=dataset.iloc[:, :-1].values
y=dataset.iloc[:,-1].values
```

```python
print(x)
```

```python
print(y)
```

## Splitting the dataset into the Training set and Test set

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=.2,random_state=0)
```

## Training the Simple Linear Regression model on the Training set

```python
from sklearn.linear_model import LinearRegression
regressor=LinearRegression()
regressor.fit(x_train,y_train)
```

## Predicting the Test set results

```python
y_pred=regressor.predict(x_test)
```

## Visualising the Training set results

```python
plt.scatter(x_train,y_train)
```

## Visualising the Test set results

```python
plt.scatter(x_test,y_test)
```

```python
plt.scatter(x_test, y_test, color='blue', label='Actual Values')
plt.plot(x_test, y_pred, color='red', label='Predicted Values', alpha=0.6)
plt.title('Comparison of Actual and Predicted Values')
plt.xlabel('Years of experience')
plt.ylabel('Salary')
plt.legend()
plt.show()
```

The alpha=0.6 parameter refers to the transparency level of the plotted points in the scatter plot. It controls how transparent or opaque the points appear:

- alpha=1.0: Fully opaque (no transparency).
- alpha=0.0: Fully transparent (invisible).
- alpha=0.6: Partially transparent (60% opaque, 40% transparent).
