```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values
```

```python
print(X)
```

```python
print(y)
```

```python
X_mean = np.mean(X, axis=0)
X_std = np.std(X, axis=0)
X = (X - X_mean) / X_std
```

```python
train_size = int(0.8 * len(X))
print(train_size)
X_train = X[:train_size]
y_train = y[:train_size]
X_test = X[train_size:]
y_test = y[train_size:]
```

```python
from sklearn.preprocessing import LabelEncoder

labelencoder = LabelEncoder()

# Encode the labels (convert categorical labels into integers)
y = labelencoder.fit_transform(y)
```

```python
# Initialize weights, bias, and hyperparameters
weights = np.zeros(X.shape[1])  # Initialize weights to zeros
bias = 0                        # Initialize bias to zero
learning_rate = 0.01            # Learning rate
iterations = 1000               # Number of iterations

# Training loop
m = X.shape[0]  # Number of samples

for i in range(iterations):
    # Step 1: Compute the linear model
    z = np.dot(X, weights) + bias

    # Step 2: Apply the sigmoid function to get predictions
    y_pred = 1 / (1 + np.exp(-z))

    # Step 3: Compute the gradients
    dw = (1 / m) * np.dot(X.T, (y_pred - y))  # Gradient w.r.t. weights
    db = (1 / m) * np.sum(y_pred - y)         # Gradient w.r.t. bias

    # Step 4: Update weights and bias
    weights -= learning_rate * dw
    bias -= learning_rate * db

    # Step 5: Compute loss for monitoring (optional)
    if i % 100 == 0:
        loss = -(1 / m) * np.sum(y * np.log(y_pred) + (1 - y) * np.log(1 - y_pred))
        print(f"Iteration {i}, Loss: {loss}")

# Making predictions
z = np.dot(X, weights) + bias
y_pred = 1 / (1 + np.exp(-z))
predictions = [1 if prob > 0.5 else 0 for prob in y_pred]

# Output the final weights, bias, and predictions
print("Final weights:", weights)
print("Final bias:", bias)
print("Predictions:", predictions)
```

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(random_state = 0)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
print(y_pred)
```

```python
print(np.concatenate((y_pred.reshape(len(y_pred),1), y_test.reshape(len(y_test),1)),1))
```

```python
accuracy = np.sum(y_pred == y_test) / len(y_test)
print("Test Accuracyis:",accuracy*100,"%")
```
