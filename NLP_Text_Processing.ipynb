```python
paragraph = input("Enter Query:")
```

```python
# lowercase
paragraph = paragraph.lower()
```

```python
import re
paragraph = re.sub(r'a-zA-Z0-9', '', paragraph)
paragraph
```

```python
import string
paragraph = list(paragraph)
print("".join(paragraph))
for i in range(1, len(paragraph) - 1):
    if paragraph[i] == '.':
        if paragraph[i-1].isdigit() and paragraph[i+1].isdigit():
            continue
        else:
            paragraph[i] = ""

if paragraph[-1] == '.' and not (paragraph[-2].isdigit() and paragraph[-1].isdigit()):
    paragraph[-1] = ""

print("".join(paragraph))
```

```python
cleaned_text = "".join(paragraph)
print(cleaned_text)
```

```python
# pos tagging
from nltk.tokenize import word_tokenize
from nltk import pos_tag
tokens = word_tokenize(cleaned_text)
pos_tags = pos_tag(tokens)
print('\n')
print(pos_tags)
print('\n')
```

```python
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
punctuations = set(string.punctuation)

# Step 3: Filter tokens
filtered_tokens = [word for word in tokens if word.lower() not in stop_words and word not in punctuations]

print(" ".join(filtered_tokens))
```

```python
from nltk.stem import PorterStemmer
ps = PorterStemmer()

stemming_word = []
for word in filtered_tokens:
  stemming_word.append(ps.stem(word))

print(stemming_word)
```

```python
from nltk.stem import WordNetLemmatizer
wl = WordNetLemmatizer()

stemming_word_lemma = []
for word in filtered_tokens:
  stemming_word_lemma.append(wl.lemmatize(word))

print(stemming_word_lemma)
```
